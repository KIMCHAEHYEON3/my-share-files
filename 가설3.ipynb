```python
# ============================================================
# 정보나루 인기대출도서 2020~2024 분석 코드
# Google Colab 실행용
# 파일명:
# 정보나루_인기대출도서_2020.csv
# 정보나루_인기대출도서_2021.csv
# 정보나루_인기대출도서_2022.csv
# 정보나루_인기대출도서_2023.csv
# 정보나루_인기대출도서_2024.csv
# ============================================================

# ------------------------------------------------------------
# 1. 라이브러리 설치 및 불러오기
# ------------------------------------------------------------

!apt-get -qq install fonts-nanum > /dev/null
!pip -q install wordcloud openpyxl

import os
import re
import ast
import shutil
import warnings
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from wordcloud import WordCloud

from google.colab import files
from IPython.display import display

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 2. 한글 폰트 설정
# ------------------------------------------------------------

font_path = "/usr/share/fonts/truetype/nanum/NanumGothic.ttf"

fm.fontManager.addfont(font_path)
plt.rcParams["font.family"] = "NanumGothic"
plt.rcParams["axes.unicode_minus"] = False

# ------------------------------------------------------------
# 3. 파일 업로드
# ------------------------------------------------------------

print("아래 5개 파일을 업로드하세요.")
print("정보나루_인기대출도서_2020.csv")
print("정보나루_인기대출도서_2021.csv")
print("정보나루_인기대출도서_2022.csv")
print("정보나루_인기대출도서_2023.csv")
print("정보나루_인기대출도서_2024.csv")

uploaded = files.upload()

# ------------------------------------------------------------
# 4. 파일 읽기 함수
# ------------------------------------------------------------

def read_csv_safely(file_name):
    encodings = ["cp949", "utf-8-sig", "utf-8", "euc-kr"]

    for enc in encodings:
        try:
            df = pd.read_csv(file_name, encoding=enc)
            return df
        except:
            continue

    raise ValueError(f"파일을 읽을 수 없습니다: {file_name}")


# ------------------------------------------------------------
# 5. 지정 파일명 기준으로 2020~2024 파일 불러오기
# ------------------------------------------------------------

expected_years = [2020, 2021, 2022, 2023, 2024]

def find_uploaded_file(year):
    base_name = f"정보나루_인기대출도서_{year}"

    # 정확히 일치하는 파일 우선
    exact_name = base_name + ".csv"
    if exact_name in uploaded.keys():
        return exact_name

    # 코랩에서 같은 파일을 여러 번 올려 (1), (2)가 붙은 경우도 처리
    for fname in uploaded.keys():
        if fname.startswith(base_name) and fname.lower().endswith(".csv"):
            return fname

    raise FileNotFoundError(f"{base_name}.csv 파일을 찾을 수 없습니다.")

dfs = []

for year in expected_years:
    file_name = find_uploaded_file(year)
    temp = read_csv_safely(file_name)
    temp["연도"] = year
    temp["원본파일명"] = file_name
    dfs.append(temp)

data = pd.concat(dfs, ignore_index=True)

print("파일 통합 완료")
print("전체 행/열:", data.shape)
print("현재 컬럼명:")
print(list(data.columns))
display(data.head())

# ------------------------------------------------------------
# 6. 네 CSV 구조에 맞춰 컬럼명 정리
# ------------------------------------------------------------

data = data.rename(columns={
    "rank": "순위",
    "title": "도서명",
    "author": "저자",
    "publisher": "출판사",
    "pub_date": "출판연도",
    "category": "주제분야",
    "is_steady": "is_steady",
    "title_clean": "title_clean",
    "keywords": "keywords",
    "KDC": "KDC 분야"
})

required_cols = ["연도", "순위", "도서명", "출판연도", "is_steady", "keywords", "KDC 분야"]

missing_cols = [col for col in required_cols if col not in data.columns]

if missing_cols:
    print("누락된 컬럼:", missing_cols)
    print("현재 컬럼:", list(data.columns))
    raise ValueError("필수 컬럼이 누락되었습니다.")

print("컬럼명 정리 완료")
print(list(data.columns))
display(data.head())

# ------------------------------------------------------------
# 7. 기본 전처리
# ------------------------------------------------------------

# 순위 숫자만 추출
def clean_rank(x):
    if pd.isna(x):
        return np.nan

    x = str(x)
    match = re.search(r"\d+", x)

    if match:
        return int(match.group())
    return np.nan

data["rank_clean"] = data["순위"].apply(clean_rank)

# 출판연도 정리
def clean_pub_year(x):
    if pd.isna(x):
        return np.nan

    x = str(x)
    match = re.search(r"\d{4}", x)

    if match:
        return int(match.group())
    return np.nan

data["출판연도_clean"] = data["출판연도"].apply(clean_pub_year)

# 스테디셀러 여부 정리
def clean_steady(x):
    if pd.isna(x):
        return False

    if isinstance(x, bool):
        return x

    x = str(x).strip().lower()

    if x in ["true", "1", "yes", "y", "o", "스테디", "스테디셀러"]:
        return True

    if x in ["false", "0", "no", "n", "x", "트렌드", "비스테디"]:
        return False

    return False

data["is_steady"] = data["is_steady"].apply(clean_steady)

# 키워드 분리
# 현재 파일의 keywords 형태: ['꿈', '백화점']
def parse_keywords(x):
    if pd.isna(x):
        return []

    if isinstance(x, list):
        return x

    x = str(x).strip()

    if x == "":
        return []

    # ['꿈', '백화점'] 형태 처리
    try:
        parsed = ast.literal_eval(x)
        if isinstance(parsed, list):
            return [str(k).strip() for k in parsed if str(k).strip() != ""]
    except:
        pass

    # 혹시 쉼표 문자열이면 처리
    parts = re.split(r"[,;/|\n]+", x)
    return [p.strip() for p in parts if p.strip() != ""]

data["keyword_list"] = data["keywords"].apply(parse_keywords)

# 결과 저장 폴더
output_dir = "information_naru_outputs"
os.makedirs(output_dir, exist_ok=True)

print("전처리 완료")
display(data.head())

# ------------------------------------------------------------
# 8. 키워드 long 데이터 생성
# ------------------------------------------------------------

keyword_rows = []

for _, row in data.iterrows():
    for kw in row["keyword_list"]:
        keyword_rows.append({
            "연도": row["연도"],
            "순위": row["순위"],
            "rank_clean": row["rank_clean"],
            "도서명": row["도서명"],
            "저자": row["저자"],
            "출판사": row["출판사"],
            "출판연도": row["출판연도_clean"],
            "주제분야": row["주제분야"],
            "KDC 분야": row["KDC 분야"],
            "is_steady": row["is_steady"],
            "키워드": kw
        })

keyword_df = pd.DataFrame(keyword_rows)

if keyword_df.empty:
    raise ValueError("키워드 데이터가 비어 있습니다. keywords 컬럼을 확인하세요.")

print("키워드 데이터 생성 완료")
display(keyword_df.head())

# ------------------------------------------------------------
# 9. 연도별 스테디셀러와 트렌드 도서 구성
# ------------------------------------------------------------

summary_rows = []

for year, group in data.groupby("연도"):
    total_books = len(group)
    steady_count = int(group["is_steady"].sum())
    trend_count = total_books - steady_count

    unique_keywords = set()
    for kws in group["keyword_list"]:
        unique_keywords.update(kws)

    summary_rows.append({
        "연도": year,
        "도서 수": total_books,
        "스테디셀러": steady_count,
        "트렌드 도서": trend_count,
        "고유 키워드 수": len(unique_keywords),
        "고유 제목 수": group["도서명"].nunique(),
        "스테디 비율(%)": round(steady_count / total_books * 100, 1),
        "트렌드 비율(%)": round(trend_count / total_books * 100, 1)
    })

steady_table = pd.DataFrame(summary_rows).sort_values("연도")

print("표 1. 연도별 스테디셀러와 트렌드 도서 구성")
display(steady_table)

plt.figure(figsize=(10, 6))

years = steady_table["연도"].astype(str)
trend_values = steady_table["트렌드 도서"]
steady_values = steady_table["스테디셀러"]

plt.bar(years, trend_values, label="트렌드 도서")
plt.bar(years, steady_values, bottom=trend_values, label="스테디셀러")

for _, row in steady_table.iterrows():
    plt.text(
        str(row["연도"]),
        row["도서 수"] + 1,
        f"{row['스테디셀러']}\n({row['스테디 비율(%)']}%)",
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.title("연도별 스테디셀러와 트렌드 도서 구성")
plt.xlabel("연도")
plt.ylabel("도서 수")
plt.legend()
plt.tight_layout()
plt.savefig(f"{output_dir}/01_steady_trend_composition.png", dpi=300)
plt.show()

# ------------------------------------------------------------
# 10. KDC 분야별 점유율 변화
# ------------------------------------------------------------

kdc_count = (
    data.groupby(["연도", "KDC 분야"])
    .size()
    .reset_index(name="도서 수")
)

kdc_count["점유율(%)"] = (
    kdc_count.groupby("연도")["도서 수"]
    .transform(lambda x: x / x.sum() * 100)
)

kdc_pivot = (
    kdc_count
    .pivot(index="연도", columns="KDC 분야", values="점유율(%)")
    .fillna(0)
)

print("KDC 분야별 점유율")
display(kdc_pivot)

plt.figure(figsize=(10, 6))

for col in kdc_pivot.columns:
    plt.plot(kdc_pivot.index, kdc_pivot[col], marker="o", label=col)

plt.title("KDC 분야별 점유율 변화")
plt.xlabel("연도")
plt.ylabel("점유율(%)")
plt.xticks(sorted(data["연도"].unique()))
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(f"{output_dir}/02_kdc_share_by_year.png", dpi=300)
plt.show()

# ------------------------------------------------------------
# 11. 제목 기반 핵심 키워드 흐름
# ------------------------------------------------------------

keyword_count = (
    keyword_df.groupby(["연도", "키워드"])
    .size()
    .reset_index(name="키워드 등장 횟수")
)

selected_keywords = [
    "설민석", "한국사", "과학", "전천당",
    "마법", "꿈", "백화점", "호기심",
    "대모험", "흔한남매"
]

selected_data = keyword_count[keyword_count["키워드"].isin(selected_keywords)]

selected_pivot = (
    selected_data
    .pivot(index="연도", columns="키워드", values="키워드 등장 횟수")
    .fillna(0)
)

print("특정 핵심 키워드의 연도별 변화")
display(selected_pivot)

plt.figure(figsize=(11, 6))

for col in selected_pivot.columns:
    plt.plot(selected_pivot.index, selected_pivot[col], marker="o", label=col)

plt.title("특정 핵심 키워드의 연도별 변화 추이")
plt.xlabel("연도")
plt.ylabel("키워드 등장 횟수")
plt.xticks(sorted(data["연도"].unique()))
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(f"{output_dir}/03_selected_keywords_trend.png", dpi=300)
plt.show()

# ------------------------------------------------------------
# 12. 스테디셀러 포함 여부에 따른 키워드 다양성
# ------------------------------------------------------------

diversity_rows = []

for year, group in data.groupby("연도"):
    all_keywords = set()
    nonsteady_keywords = set()
    steady_keywords = set()

    for _, row in group.iterrows():
        kws = row["keyword_list"]
        all_keywords.update(kws)

        if row["is_steady"]:
            steady_keywords.update(kws)
        else:
            nonsteady_keywords.update(kws)

    diversity_rows.append({
        "연도": year,
        "전체 고유 키워드 수": len(all_keywords),
        "스테디셀러 제외 키워드 수": len(nonsteady_keywords),
        "스테디셀러 키워드 수": len(steady_keywords)
    })

diversity_table = pd.DataFrame(diversity_rows).sort_values("연도")

print("표 2. 스테디셀러 포함 여부에 따른 키워드 다양성")
display(diversity_table)

plt.figure(figsize=(10, 6))

plt.plot(
    diversity_table["연도"],
    diversity_table["전체 고유 키워드 수"],
    marker="o",
    label="전체 고유 키워드 수"
)

plt.plot(
    diversity_table["연도"],
    diversity_table["스테디셀러 제외 키워드 수"],
    marker="o",
    label="스테디셀러 제외 키워드 수"
)

plt.plot(
    diversity_table["연도"],
    diversity_table["스테디셀러 키워드 수"],
    marker="o",
    label="스테디셀러 키워드 수"
)

plt.title("스테디셀러 포함 여부에 따른 키워드 다양성 변화")
plt.xlabel("연도")
plt.ylabel("고유 키워드 수")
plt.xticks(sorted(data["연도"].unique()))
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(f"{output_dir}/04_keyword_diversity_by_steady.png", dpi=300)
plt.show()

# ------------------------------------------------------------
# 13. 인접 연도 간 키워드 Jaccard 유사도
# ------------------------------------------------------------

year_keywords = {}

for year, group in data.groupby("연도"):
    kws = set()

    for keyword_list in group["keyword_list"]:
        kws.update(keyword_list)

    year_keywords[year] = kws

jaccard_rows = []
years_sorted = sorted(year_keywords.keys())

for i in range(len(years_sorted) - 1):
    y1 = years_sorted[i]
    y2 = years_sorted[i + 1]

    set1 = year_keywords[y1]
    set2 = year_keywords[y2]

    common = set1 & set2
    union = set1 | set2

    score = len(common) / len(union) if len(union) > 0 else 0

    jaccard_rows.append({
        "비교구간": f"{y1}-{y2}",
        "공통 키워드 수": len(common),
        "통합 키워드 수": len(union),
        "Jaccard 유사도": round(score, 2)
    })

jaccard_table = pd.DataFrame(jaccard_rows)

print("표 3. 인접 연도 간 키워드 Jaccard 유사도")
display(jaccard_table)

plt.figure(figsize=(10, 6))

plt.plot(
    jaccard_table["비교구간"],
    jaccard_table["Jaccard 유사도"],
    marker="o"
)

for _, row in jaccard_table.iterrows():
    plt.text(
        row["비교구간"],
        row["Jaccard 유사도"] + 0.02,
        f"{row['Jaccard 유사도']:.2f}",
        ha="center"
    )

plt.title("인접 연도 간 키워드 Jaccard 유사도")
plt.xlabel("비교구간")
plt.ylabel("Jaccard 유사도")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(f"{output_dir}/05_jaccard_similarity.png", dpi=300)
plt.show()

# ------------------------------------------------------------
# 14. 순위가중점수 기반 보조 분석
# ------------------------------------------------------------

keyword_df["순위가중점수"] = 101 - keyword_df["rank_clean"]
keyword_df["순위가중점수"] = keyword_df["순위가중점수"].clip(lower=0)

weighted_score = (
    keyword_df.groupby(["키워드", "연도"])["순위가중점수"]
    .sum()
    .reset_index()
)

top_keywords = (
    weighted_score.groupby("키워드")["순위가중점수"]
    .sum()
    .sort_values(ascending=False)
    .head(12)
    .index
)

heatmap_data = (
    weighted_score[weighted_score["키워드"].isin(top_keywords)]
    .pivot(index="키워드", columns="연도", values="순위가중점수")
    .fillna(0)
)

heatmap_data["합계"] = heatmap_data.sum(axis=1)
heatmap_data = heatmap_data.sort_values("합계", ascending=False)
heatmap_data = heatmap_data.drop(columns="합계")

print("순위가중점수 기반 상위 키워드")
display(heatmap_data)

plt.figure(figsize=(10, 7))

plt.imshow(heatmap_data, aspect="auto")

plt.xticks(range(len(heatmap_data.columns)), heatmap_data.columns)
plt.yticks(range(len(heatmap_data.index)), heatmap_data.index)

plt.colorbar(label="순위가중점수")
plt.title("순위가중점수 기반 상위권 집중 키워드 지도")
plt.xlabel("연도")
plt.ylabel("키워드")

for i in range(len(heatmap_data.index)):
    for j in range(len(heatmap_data.columns)):
        value = int(heatmap_data.iloc[i, j])

        if value > 0:
            plt.text(j, i, value, ha="center", va="center", fontsize=8)

plt.tight_layout()
plt.savefig(f"{output_dir}/06_rank_weighted_heatmap.png", dpi=300)
plt.show()

# ------------------------------------------------------------
# 15. 2020~2024 전체 WordCloud
# ------------------------------------------------------------

all_keyword_counter = Counter(keyword_df["키워드"].dropna().tolist())

wordcloud = WordCloud(
    font_path=font_path,
    width=1600,
    height=800,
    background_color="white",
    max_words=200,
    random_state=42
).generate_from_frequencies(all_keyword_counter)

plt.figure(figsize=(14, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("2020~2024 전체 핵심 키워드 WordCloud")
plt.tight_layout()
plt.savefig(f"{output_dir}/07_wordcloud_all.png", dpi=300)
plt.show()

# 제목 없는 전체 WordCloud도 저장
plt.figure(figsize=(14, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.tight_layout()
plt.savefig(f"{output_dir}/07_wordcloud_all_no_title.png", dpi=300)
plt.show()

# ------------------------------------------------------------
# 16. 연도별 WordCloud 개별 저장
# ------------------------------------------------------------

for year in years_sorted:
    year_counter = Counter(
        keyword_df[keyword_df["연도"] == year]["키워드"].dropna().tolist()
    )

    wc = WordCloud(
        font_path=font_path,
        width=900,
        height=600,
        background_color="white",
        max_words=100,
        random_state=42
    ).generate_from_frequencies(year_counter)

    plt.figure(figsize=(8, 5))
    plt.imshow(wc, interpolation="bilinear")
    plt.axis("off")
    plt.tight_layout()
    plt.savefig(f"{output_dir}/08_wordcloud_{year}_no_title.png", dpi=300)
    plt.show()

# ------------------------------------------------------------
# 17. 연도별 WordCloud 모음 저장
# ------------------------------------------------------------

fig, axes = plt.subplots(1, len(years_sorted), figsize=(18, 4))

if len(years_sorted) == 1:
    axes = [axes]

for ax, year in zip(axes, years_sorted):
    year_counter = Counter(
        keyword_df[keyword_df["연도"] == year]["키워드"].dropna().tolist()
    )

    wc = WordCloud(
        font_path=font_path,
        width=600,
        height=500,
        background_color="white",
        max_words=100,
        random_state=42
    ).generate_from_frequencies(year_counter)

    ax.imshow(wc, interpolation="bilinear")
    ax.axis("off")
    ax.set_title(f"{year}년")

plt.tight_layout()
plt.savefig(f"{output_dir}/09_wordcloud_by_year_collection.png", dpi=300)
plt.show()

# 제목 없는 연도별 WordCloud 모음 저장
fig, axes = plt.subplots(1, len(years_sorted), figsize=(18, 4))

if len(years_sorted) == 1:
    axes = [axes]

for ax, year in zip(axes, years_sorted):
    year_counter = Counter(
        keyword_df[keyword_df["연도"] == year]["키워드"].dropna().tolist()
    )

    wc = WordCloud(
        font_path=font_path,
        width=600,
        height=500,
        background_color="white",
        max_words=100,
        random_state=42
    ).generate_from_frequencies(year_counter)

    ax.imshow(wc, interpolation="bilinear")
    ax.axis("off")

plt.tight_layout()
plt.savefig(f"{output_dir}/09_wordcloud_by_year_collection_no_title.png", dpi=300)
plt.show()

# ------------------------------------------------------------
# 18. 분석 결과 표 저장
# ------------------------------------------------------------

steady_table.to_csv(
    f"{output_dir}/table_01_steady_trend.csv",
    index=False,
    encoding="utf-8-sig"
)

diversity_table.to_csv(
    f"{output_dir}/table_02_keyword_diversity.csv",
    index=False,
    encoding="utf-8-sig"
)

jaccard_table.to_csv(
    f"{output_dir}/table_03_jaccard_similarity.csv",
    index=False,
    encoding="utf-8-sig"
)

kdc_pivot.to_csv(
    f"{output_dir}/table_04_kdc_share.csv",
    encoding="utf-8-sig"
)

selected_pivot.to_csv(
    f"{output_dir}/table_05_selected_keywords.csv",
    encoding="utf-8-sig"
)

heatmap_data.to_csv(
    f"{output_dir}/table_06_rank_weighted_score.csv",
    encoding="utf-8-sig"
)

keyword_count.to_csv(
    f"{output_dir}/table_07_keyword_count.csv",
    index=False,
    encoding="utf-8-sig"
)

data.to_csv(
    f"{output_dir}/정보나루_인기대출도서_2020_2024_통합.csv",
    index=False,
    encoding="utf-8-sig"
)

keyword_df.to_csv(
    f"{output_dir}/정보나루_키워드_long_data.csv",
    index=False,
    encoding="utf-8-sig"
)

# ------------------------------------------------------------
# 19. 결과 압축 및 다운로드
# ------------------------------------------------------------

zip_name = "information_naru_analysis_outputs"

if os.path.exists(zip_name + ".zip"):
    os.remove(zip_name + ".zip")

shutil.make_archive(zip_name, "zip", output_dir)

print("분석 완료")
print("결과 이미지와 표가 압축 파일로 저장되었습니다.")

files.download(zip_name + ".zip")
```
